```python


def max_profit(n):
    max_earnings = 0
    solutions = []

    for t in range(n // 5 + 1):
        for p in range(n // 4 + 1):
            for c in range(n // 10 + 1):

                buildings = (
                    [('T', 5, 1500)] * t
                    + [('P', 4, 1000)] * p
                    + [('C', 10, 2000)] * c
                )

                buildings.sort(
                    key=lambda b: b[2] / b[1],
                    reverse=True
                )

                time_used = 0
                earnings = 0

                for _, build_time, rate in buildings:
                    time_used += build_time
                    earnings += rate * (n - time_used)

                if time_used > n:
                    continue

                result = (t, p, c)

                if earnings > max_earnings:
                    max_earnings = earnings
                    solutions = [result]
                elif earnings == max_earnings:
                    solutions.append(result)

    print("Output Earnings: $" + str(max_earnings))
    print("Solutions")

    for i, (t, p, c) in enumerate(solutions, 1):
        print(f"{i}. T: {t} P: {p} C: {c}")


n = int(input("Input Time Unit: "))
max_profit(n)
```
